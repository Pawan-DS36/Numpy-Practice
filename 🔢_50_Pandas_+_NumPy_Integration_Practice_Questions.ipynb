💡 **Hint:** Use `time.time()` before and after each operation. Apply the same computation (`arr * 2 + 1`) to a raw NumPy array and a Pandas Series. NumPy is typically faster on pure numeric operations.

### NP34. Calculate variance and standard deviation.
> 💡 **Hint:** `np.var(arr)` and `np.std(arr)` use `ddof=0` (population). Pandas `.var()` and `.std()` use `ddof=1` (sample). Be aware of this difference when comparing results.

```python
arr = df["col1"].values

# 1. NumPy Calculations (Default: Population, ddof=0)
np_var = np.var(arr)
np_std = np.std(arr)

# 2. Pandas Calculations (Default: Sample, ddof=1)
pd_var = df["col1"].var()
pd_std = df["col1"].std()

print(f"NumPy (Population) -> Var: {np_var}, Std: {np_std}")
print(f"Pandas (Sample)     -> Var: {pd_var}, Std: {pd_std}")

# 3. Matching them up by adjusting ddof
print("\n--- Matching the Results ---")
print(f"NumPy with ddof=1 (Sample):     {np.std(arr, ddof=1)}")
print(f"Pandas with ddof=0 (Population): {df['col1'].std(ddof=0)}")
```

```python
data = {
    "col1": [10, 12, 11, 15, 100, 12, 14, 13, 11, 9],
    "col2": [5, 15, 25, 35, 45, 55, 65, 75, 85, 95],
    "category": ["DS", "ML", "AI", "DS", "ML", "AI", "DS", "ML", "AI", "DS"]
}
df = pd.DataFrame(data)

arr = df["col1"].values
q1,q3 = np.percentile(arr,[25,75])

iqr = q3 - q1
lower_bound = q1 - 1.5*iqr
upper_bound = q3 + 1.5*iqr

df["col1_cleaned"] = np.clip(arr,lower_bound,upper_bound)
print("lower",lower_bound)
print("upper",upper_bound)
print(df[['col1','col1_cleaned']])
```

### NP49. Perform a mini data analysis project.
> 💡 **Hint:** Follow these steps: (1) Generate data with NumPy, (2) Store in DataFrame, (3) Clean with Pandas, (4) Engineer features with NumPy, (5) GroupBy analysis, (6) Detect outliers, (7) Export summary.

```python
raw_met = np.random.normal(loc=200,scale=50,size=15)

pro_df = pd.DataFrame({"Revenue" : raw_met})
pro_df["Region"] = np.random.choice(["North","South","East"],size=15)

pro_df["Revenue"] = pro_df["Revenue"].fillna(pro_df["Revenue"].median())

pro_df["Log_Revenue"] = np.log1p(pro_df["Revenue"])

summ = pro_df.groupby("Region")["Revenue"].agg(["mean","std","count"])
print(summ)

q25,q75 = np.percentile(pro_df["Revenue"],[25,75])
pro_df["Is_Outlier"] = ~pro_df["Revenue"].between(q25 - 1.5*(q75-q25),q75 + 1.5*(q75-q25))
print(pro_df[["Revenue","Is_Outlier"]])

pro_df.to_csv("mini_analysis_output.csv", index=False)
print("\nMini analysis pipeline exported successfully.")
```

### NP50. Build an end-to-end pipeline using NumPy + Pandas.
> 💡 **Hint:** Chain all steps together:  
> 1. `np.random.seed(X)` → generate raw data  
> 2. `np.clip` → clean invalid values  
> 3. `np.log`, `np.square`, `np.where` → feature engineering  
> 4. `df.groupby().agg(...)` → statistical analysis  
> 5. IQR + `np.clip` → outlier removal  
> 6. `df[features].to_numpy(dtype=np.float32)` → ML feature matrix  
> 7. Export with `df.to_csv(...)`

```python
def run_pipeline(seed_value=42):
  np.random.seed(seed_value)
  rows = 200

  raw_data = {
      "Age" : np.random.uniform(low=15 ,high=90,size=rows),
      "Income" : np.random.normal(loc=55000,scale=25000,size=rows),
      "score" : np.random.choice([1,2,3,4],size = rows)
  }
  pp_df = pd.DataFrame(raw_data)

  pp_df["Age"] = np.clip(pp_df["Age"],18,80)

  pp_df["Log_Income"] = np.log1p(np.maximum(0,pp_df["Income"]))
  pp_df["Age_Squared"] = np.square(pp_df["Age"])
  pp_df["High_Earner"] = np.where(pp_df["Income"] > 60000,1,0)

  summ = pp_df.groupby("score")["Age"].agg(["mean","median"])
  print("Pipeline grouped checkpoint",sum)

  in_vals = pp_df["Income"].values
  q1_in, q3_in = np.percentile(in_vals,[25,75])
  iqr_in = q3_in - q1_in
  pp_df["Income_cleaned"] = np.clip(in_vals,q1_in - 1.5 * iqr_in,q3_in + 1.5 * iqr_in)

  ml_features = ["Age","Income_cleaned","Log_Income","Age_Squared","High_Earner"]
  Xmat = pp_df[ml_features].to_numpy(dtype=np.float32)
  print("\nFinal Matrix Export shape",Xmat.shape)

  pp_df.to_csv("final_pipeline_dataset.csv",index=False)
  print("Pipeline process run finished successfully.")

run_pipeline()
```
