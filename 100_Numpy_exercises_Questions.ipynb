#### 80. Consider an arbitrary array, write a function that extracts a subpart with a fixed shape and centered on a given element (pad with a `fill` value when necessary) (★★★)
`hint: minimum maximum`

````python
def extract(arr, shape, position, fill=0):
    R = np.ones(shape, dtype=arr.dtype) * fill
    P = np.array(list(position))
    S = np.array(list(shape))
    start = np.maximum(0, P - S // 2)
    end   = np.minimum(arr.shape, P + S // 2 + 1)
    rs    = tuple(slice(s, e) for s, e in zip(start, end))
    rs2   = tuple(slice(s, e) for s, e in zip(start - P + S // 2, end - P + S // 2))
    R[rs2] = arr[rs]
    return R
arr = np.arange(25).reshape(5, 5)
print(extract(arr, (3, 3), (2, 2)))
````

````python
class SymmetricArray(np.ndarray):
    def __setitem__(self, idx, val):
        i, j = idx
        super().__setitem__((i, j), val)
        super().__setitem__((j, i), val)


arr = np.zeros((5, 5)).view(SymmetricArray)
arr[1, 3] = 9
print(arr[3, 1])
````

#### 88. How to implement the Game of Life using numpy arrays? (★★★)
`No hints provided...`

````python
def game_of_life(Z):
    N = sum(np.roll(np.roll(Z, i, 0), j, 1)
            for i in (-1, 0, 1) for j in (-1, 0, 1) if (i != 0 or j != 0))
    return (N == 3) | (Z & (N == 2))
Z = np.random.randint(0, 2, (10, 10))
print(game_of_life(Z))
````

#### 98. Considering a path described by two vectors (X,Y), how to sample it using equidistant samples (★★★)?
`hint: np.cumsum, np.interp`

````python
X = np.array([0, 1, 2, 3, 4], dtype=float)
Y = np.array([0, 1, 0, 1, 0], dtype=float)
dists = np.cumsum(np.sqrt(np.diff(X)**2 + np.diff(Y)**2))
dists = np.insert(dists, 0, 0)
samples = np.linspace(0, dists[-1], 10)
Xi = np.interp(samples, dists, X)
Yi = np.interp(samples, dists, Y)
print(Xi, Yi)
````

````python
X = np.random.randint(0, 5, (10, 3))
n = 4
rows = X[np.logical_and.reduce([X.sum(axis=1) == n,
                                 np.mod(X, 1) == 0], axis=0)]
print(rows)
````

#### 100. Compute bootstrapped 95% confidence intervals for the mean of a 1D array X (i.e., resample the elements of an array with replacement N times, compute the mean of each sample, and then compute percentiles over the means). (★★★)
`hint: np.percentile`

````python
X = np.random.random(100)
N = 1000
means = np.array([np.random.choice(X, len(X), replace=True).mean() for _ in range(N)])
lower, upper = np.percentile(means, [2.5, 97.5])
print("95% CI:", lower, "to", upper)
````
